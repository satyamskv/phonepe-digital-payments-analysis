```python
#1.3 Checking missing Values

#1.3.1 Varifying missing values availability in dataset


for original_table_name, table in all_tables.items():
    print("-" * 100)
    missing = table.isnull().sum()
    missing = missing[missing > 0]
    print("Missing values in:", original_table_name)
    if len(missing) >0:
        print(missing)
    else:
        print("No missing values")

    print()
```

```python
#1.3.2 Percentage of missing values
for original_table_name, table in all_tables.items():
    print("-"*100)
    missing_percent = (table.isnull().sum() / len(table)) * 100
    missing_percent = missing_percent[missing_percent > 0].sort_values(ascending = False).round(2)
    if len(missing_percent) > 0:
        print(original_table_name, "have",missing_percent)
    else:
        print(original_table_name, "have no missing values")
    print()
```

```python
#1.3.2 Column having the highest percentage of missing values
for original_table_name, table in all_tables.items():
    print()
    print("-" * 100)
    print()
    missing_percent = (table.isnull().sum() / len(table)) * 100
    if missing_percent.max() > 0:
        highest_percent = missing_percent.idxmax()
        print(original_table_name, highest_percent, round(missing_percent.max(), 2), '% missing')
    else:
        print(original_table_name, "Has no missing values")
```

```python
#3 Data Quality Checks
#3.1.1 For each state, the total number of transactions, total transaction amount, and total registered users from district level data

district_level_totals = district_data.groupby("State")[["Transactions", "Amount (INR)", "Registered Users"]].sum()
district_level_totals.head(10)
```

```python
#3.1.2 Comparing the values with state level to check if they match
state_level_totals = state_data.groupby("State")[["Transactions", "Amount (INR)", "Registered Users"]].sum().reset_index()
district_level_totals = district_data.groupby("State")[["Transactions", "Amount (INR)", "Registered Users"]].sum().reset_index()

comparison = state_level_totals.merge(district_level_totals, on="State", suffixes=("_state", "_district"), how="outer")

comparison["Transactions diff"] = comparison["Transactions_state"] - comparison["Transactions_district"]
comparison["Amount diff"] = comparison["Amount (INR)_state"] - comparison["Amount (INR)_district"]
comparison["Users diff"] = comparison["Registered Users_state"] - comparison["Registered Users_district"]

comparison[["Transactions diff", "Amount diff", "Users diff"]]
```

```python
#3.1.3 Displaying any discrepancies found between district-level and state-level data
mismatch = comparison[(comparison["Transactions diff"].abs() > 1) |
                      (comparison["Amount diff"].abs() > 1) |
                      (comparison["Users diff"].abs() > 1)]

if len(mismatch) > 0:
    print("States with discrepancies:")
    display(mismatch[["State",
                      "Transactions_state", "Transactions_district", "Transactions diff",
                      "Amount (INR)_state", "Amount (INR)_district", "Amount diff",
                      "Registered Users_state", "Registered Users_district", "Users diff"]])
else:
    print("No discrepancies found.")
```

```python
#4 Data merging and Advance Analysis
#4.1 Ratio of users to population by State

latest_year = state_data["Year"].max()
latest_quarter = state_data.loc[state_data["Year"] == latest_year, "Quarter"].max()

latest_users = state_data[(state_data["Year"] == latest_year) &
                          (state_data["Quarter"] == latest_quarter)]
latest_users = latest_users.set_index("State")["Registered Users"]

population_by_state = demographics_data.groupby("State")["Population"].sum()

users_vs_population = pd.DataFrame({
    "Registered Users": latest_users,
    "Population": population_by_state
})

users_vs_population["Users to Population Ratio"] = users_vs_population["Registered Users"] / users_vs_population["Population"]
users_vs_population = users_vs_population.sort_values("Users to Population Ratio", ascending=False)

users_vs_population
```

```python
#4.1.2 Column chart of the users-to-population ratio by state
plt.figure(figsize=(15, 6))
sns.barplot(data=users_vs_population.reset_index(), x="State",
            y="Users to Population Ratio", color="#08FFE2")

plt.xticks(rotation=90)
plt.title("Registered users as a share of population, by state")
plt.xlabel("State")
plt.ylabel("Users / Population")
plt.tight_layout()
plt.show()
```

```python
#4.2 Correlation between population density and transaction volume
#4.2.1 Merging district transactions with district demographics

district_transactions = district_data.groupby(["State", "District"])["Transactions"].sum().reset_index()
district_transactions = district_transactions.rename(columns={"Transactions": "Total Transactions"})

district_merged = district_transactions.merge(demographics_data[["State", "District", "Density", "Population"]], on=["State", "District"], how="inner")
district_merged
```

```python
#4.3 Average Transaction amount per user
#4.3.1 Average Transaction amount per user, for each state
# Note: "Registered Users" is a cumulative total, so latest_users is the state's total
# user base; dividing the overall amount by it gives the average amount per user.

total_amount = state_data.groupby("State")["Amount (INR)"].sum()
amount_per_user = pd.DataFrame({"Total Amount (INR)": total_amount,"Registered Users": latest_users})
amount_per_user["Average Amount per User"] = amount_per_user["Total Amount (INR)"] / amount_per_user["Registered Users"]
amount_per_user = amount_per_user.sort_values("Average Amount per User", ascending=False)
amount_per_user
```

```python
# 6.3 Summary of findings and insights

first_transactions = national_trend["Transactions"].iloc[0]
last_transactions  = national_trend["Transactions"].iloc[-1]
growth_percent = (last_transactions / first_transactions - 1) * 100

state_totals = state_data.groupby("State")[["Transactions", "Amount (INR)"]].sum()
state_totals = state_totals.sort_values("Transactions", ascending=False)
top_state = state_totals.index[0]
lowest_state = state_totals.index[-1]
top5_share = state_totals["Transactions"].head(5).sum() / state_totals["Transactions"].sum() * 100

type_summary = split_data.groupby("Transaction Type")[["Transactions", "Amount (INR)"]].sum()
type_summary["Share %"] = type_summary["Transactions"] / type_summary["Transactions"].sum() * 100
type_summary["ATV"] = type_summary["Amount (INR)"] / type_summary["Transactions"]
type_summary = type_summary.sort_values("Transactions", ascending=False)
most_common_type = type_summary.index[0]

latest_year = state_data["Year"].max()
latest_quarter = state_data.loc[state_data["Year"] == latest_year, "Quarter"].max()
latest_users = state_data[(state_data["Year"] == latest_year) &(state_data["Quarter"] == latest_quarter)].set_index("State")["Registered Users"]
population_by_state = demographics_data.groupby("State")["Population"].sum()
users_to_population = (latest_users / population_by_state).sort_values(ascending=False)
top_penetration_state = users_to_population.index[0]

latest_devices = device_data[(device_data["Year"] == latest_year) & (device_data["Quarter"] == latest_quarter)]
latest_devices = latest_devices.sort_values("Registered Users", ascending=False)
leading_brand = latest_devices.drop_duplicates(subset=["State"], keep="first")
brand_counts = leading_brand["Brand"].value_counts()
top_brand = brand_counts.index[0]
top_brand_states = brand_counts.iloc[0]

density_corr = round(district_merged["Density"].corr(district_merged["Total Transactions"]), 3)
population_corr = round(district_merged["Population"].corr(district_merged["Total Transactions"]), 3)


print("=" * 70)
print("KEY FINDINGS")
print("=" * 70)
print(f"1. Growth: transactions grew about {growth_percent:.0f}% between "
      f"{national_trend['Period'].iloc[0]} and {national_trend['Period'].iloc[-1]} "
      f"(from {first_transactions:,} to {last_transactions:,}).")
print(f"2. Concentration: {top_state} has the most transactions, and the top 5 states make up "
      f"about {top5_share:.0f}% of all transactions. {lowest_state} has the fewest.")
print(f"3. Transaction types: '{most_common_type}' is the most used type, with the highest "
      f"average value (Rs {type_summary['ATV'].iloc[0]:,.0f}).")
print(f"4. Penetration: {top_penetration_state} has the highest registered-users-to-population "
      f"ratio (about {users_to_population.iloc[0]*100:.0f}%).")
print(f"5. Devices: {top_brand} is the leading brand in {top_brand_states} of "
      f"{leading_brand['State'].nunique()} states.")
print(f"6. Demographics: population density vs transactions correlation = {density_corr}; "
      f"population vs transactions = {population_corr}.")

print()
print("Transaction type breakdown:")
print(type_summary[["Transactions", "Share %", "ATV"]].round(2))

print()
print("=" * 70)
print("RECOMMENDATIONS")
print("=" * 70)
print(f"- Focus infrastructure and offers on the high-volume states (like {top_state}), "
      f"where most transactions already happen.")
print(f"- Run awareness and cashback campaigns in low-volume states (like {lowest_state}) "
      f"to reduce the gap between states.")
print(f"- '{most_common_type}' is both the most used and highest-value type, so keep improving "
      f"that experience first.")
print(f"- Target low-penetration states (low users-to-population) for new user sign-ups.")
print(f"- Make sure the app performs well on the most common phone brands (like {top_brand}).")
print(f"- Use district population density as a quick guide when planning where to launch new features.")
```
